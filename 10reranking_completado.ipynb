# Ejercicio 10: Re-ranking

**Objetivo:** Implementar y evaluar un pipeline de Recuperación de Información en dos etapas, y analizar el impacto del re-ranking en la calidad del ranking.

## Parte 1. Preparación del corpus

* Cargar el corpus (documentos/pasajes).
* Cargar las consultas (queries).
* Cargar qrels (relevancia).

```python
from beir import util
from beir.datasets.data_loader import GenericDataLoader
import pandas as pd
```

```python
DATASET_NAME = "scifact"
DATA_DIR = "../data/beir_datasets"
url = f"https://public.ukp.informatik.tu-darmstadt.de/thakur/BEIR/datasets/{DATASET_NAME}.zip"
util.download_and_unzip(url, DATA_DIR)
```

```python
dataset_path = DATA_DIR + "/" + DATASET_NAME
corpus, queries, qrels = GenericDataLoader(dataset_path).load(split="test")
```

```python
df_corpus = (
    pd.DataFrame.from_dict(corpus, orient="index")
      .reset_index()
      .rename(columns={"index": "doc_id"})
)

df_corpus
```

```python
df_queries = (
    pd.DataFrame.from_dict(queries, orient="index", columns=["query"])
      .reset_index()
      .rename(columns={"index": "query_id"})
)

df_queries
```

```python
rows = []
for qid, docs in qrels.items():
    for doc_id, rel in docs.items():
        rows.append({
            "query_id": qid,
            "doc_id": doc_id,
            "relevance": rel
        })

df_qrels = pd.DataFrame(rows)
df_qrels
```

```python
# Elegimos una query cualquiera que tenga varios documentos relevantes
qid = "133"

print("Query:")
print(df_queries.loc[df_queries["query_id"] == qid, "query"].values[0])

print("\nDocumentos relevantes para esta query:")
df_qrels[(df_qrels["query_id"] == qid) & (df_qrels["relevance"] > 0)]
```

## Parte 2. Retrieval inicial (baseline)

* Implementar retrieval inicial con BM25
* Obtener métricas: Recall@10 nDCG@10

```python

from rank_bm25 import BM25Okapi
from beir.retrieval.evaluation import EvaluateRetrieval
import numpy as np

tokenized_corpus = [
    (str(row["title"]) + " " + str(row["text"])).lower().split()
    for _, row in df_corpus.iterrows()
]

bm25 = BM25Okapi(tokenized_corpus)

results_bm25 = {}
for qid, query in queries.items():
    scores = bm25.get_scores(query.lower().split())
    ranking = np.argsort(scores)[::-1]
    results_bm25[qid] = {
        df_corpus.iloc[idx]["doc_id"]: float(scores[idx])
        for idx in ranking[:100]
    }

ndcg, _map, recall, precision = EvaluateRetrieval.evaluate(
    qrels, results_bm25, [10]
)

print("BM25 baseline")
print("nDCG@10:", ndcg["NDCG@10"])
print("Recall@10:", recall["Recall@10"])
```

## Parte 3. Implementación del re-ranking _cross-encoder_

* Re-rankear los top-k candidatos para cada query.
* Identificar qué documentos cambian de posición en el top 10

```python

from sentence_transformers import CrossEncoder

cross_encoder = CrossEncoder("cross-encoder/ms-marco-MiniLM-L-6-v2")

results_cross = {}

for qid, query in queries.items():
    candidates = list(results_bm25[qid].keys())[:50]

    pairs = []
    for doc_id in candidates:
        row = df_corpus[df_corpus["doc_id"] == doc_id].iloc[0]
        doc_text = str(row["title"]) + " " + str(row["text"])
        pairs.append([query, doc_text])

    scores = cross_encoder.predict(pairs)

    reranked = sorted(
        zip(candidates, scores),
        key=lambda x: x[1],
        reverse=True
    )

    results_cross[qid] = {doc: float(score) for doc, score in reranked}

sample_qid = list(results_cross.keys())[0]

print("Top 10 BM25")
print(list(results_bm25[sample_qid].keys())[:10])

print("\nTop 10 CrossEncoder")
print(list(results_cross[sample_qid].keys())[:10])
```

## Parte 4. Implementación del re-ranking _LTR_

* Re-rankear los top-k candidatos para cada query.
* Identificar qué documentos cambian de posición en el top 10

```python

from sklearn.ensemble import RandomForestRegressor
import pandas as pd

train_rows = []

for qid, docs in qrels.items():
    query = queries[qid]

    for doc_id, bm25_score in list(results_bm25[qid].items())[:50]:
        rel = docs.get(doc_id, 0)

        row = df_corpus[df_corpus["doc_id"] == doc_id].iloc[0]
        doc_text = str(row["title"]) + " " + str(row["text"])

        train_rows.append({
            "bm25_score": bm25_score,
            "doc_length": len(doc_text.split()),
            "relevance": rel,
            "query_id": qid,
            "doc_id": doc_id
        })

ltr_df = pd.DataFrame(train_rows)

X = ltr_df[["bm25_score", "doc_length"]]
y = ltr_df["relevance"]

ltr_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
ltr_model.fit(X, y)

results_ltr = {}

for qid, query in queries.items():
    rows = []

    for doc_id, bm25_score in list(results_bm25[qid].items())[:50]:
        row = df_corpus[df_corpus["doc_id"] == doc_id].iloc[0]
        doc_text = str(row["title"]) + " " + str(row["text"])

        rows.append({
            "doc_id": doc_id,
            "bm25_score": bm25_score,
            "doc_length": len(doc_text.split())
        })

    tmp = pd.DataFrame(rows)
    preds = ltr_model.predict(tmp[["bm25_score", "doc_length"]])

    ranking = sorted(
        zip(tmp["doc_id"], preds),
        key=lambda x: x[1],
        reverse=True
    )

    results_ltr[qid] = {doc: float(score) for doc, score in ranking}

print("Top 10 LTR")
print(list(results_ltr[sample_qid].keys())[:10])
```

## Parte 5. Evaluación post re-ranking

Calcular métricas:
* nDCG@10
* MAP
* Recall@10

```python

print("=== Evaluación CrossEncoder ===")
ndcg_ce, map_ce, recall_ce, _ = EvaluateRetrieval.evaluate(
    qrels, results_cross, [10]
)

print("nDCG@10:", ndcg_ce["NDCG@10"])
print("MAP@10:", map_ce["MAP@10"])
print("Recall@10:", recall_ce["Recall@10"])

print("\n=== Evaluación LTR ===")
ndcg_ltr, map_ltr, recall_ltr, _ = EvaluateRetrieval.evaluate(
    qrels, results_ltr, [10]
)

print("nDCG@10:", ndcg_ltr["NDCG@10"])
print("MAP@10:", map_ltr["MAP@10"])
print("Recall@10:", recall_ltr["Recall@10"])
```
